```python
import json
```

```python
people_string = '''
{
  "people": [
    {
      "name": "John Smith",
      "phone": "615-555-7164",
      "emails": ["johnsmith@bogusemail.com", "john.smith@work-place.com"],
      "has_license": false
    },
    {
      "name": "Jane Doe",
      "phone": "560-555-5153",
      "emails": null,
      "has_license": true
    }
  ]
}
'''
```

```python
people_string
```

```python
data = json.loads(people_string)
data
```

```python
type(data)
```

```python
type(data['people'])
```

```python
for person in data['people']:
    print(person)
```

```python
for person in data['people']:
    print(person['name'])
```

```python
for person in data['people']:
    del person['phone']
new_string = json.dumps(data)
print(new_string)
```

```python
new_string = json.dumps(data, indent=2) # add indent to format readable
print(new_string)
```

```python
new_string = json.dumps(data, indent=2, sort_keys=True) # sort the keys in json 'email', 'has_license', 'name', ...
print(new_string)
```
